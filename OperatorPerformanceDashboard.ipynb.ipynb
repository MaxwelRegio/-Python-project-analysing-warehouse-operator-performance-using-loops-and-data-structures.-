## 📊 Operator Performance Dashboard (Python Project)

**Goal: Analyse and visualise operator productivity using loops, conditions, and basic data structures.**

🧠 Key Python Concepts Used:
for loops

if/elif/else conditions

dictionaries and lists

basic math operations

sorting and filtering

```python
# Sample data: Operator performance metrics
operators = [
    {"name": "John Murphy", "work_time": 480, "lines_actioned": 120, "quantity_actioned": 900},
    {"name": "Aisha Khan", "work_time": 450, "lines_actioned": 150, "quantity_actioned": 1100},
    {"name": "Carlos Mendes", "work_time": 500, "lines_actioned": 100, "quantity_actioned": 700},
    {"name": "Emily Zhang", "work_time": 470, "lines_actioned": 130, "quantity_actioned": 950},
    {"name": "David O'Connor", "work_time": 460, "lines_actioned": 160, "quantity_actioned": 1200},
]

# Analyze performance
for op in operators:
    lines_per_hour = round(op["lines_actioned"] / (op["work_time"] / 60), 2)
    qty_per_hour = round(op["quantity_actioned"] / (op["work_time"] / 60), 2)

    print(f"Operator: {op['name']}")
    print(f"Lines/hr: {lines_per_hour}, Qty/hr: {qty_per_hour}")

    if qty_per_hour >= 140:
        print("🏆 Top Performer")
    elif qty_per_hour >= 100:
        print("✅ Solid Performance")
    else:
        print("⚠️ Needs Support")

    print("-" * 40)

# Calculate team averages
total_lines = sum(op["lines_actioned"] for op in operators)
total_qty = sum(op["quantity_actioned"] for op in operators)
total_time = sum(op["work_time"] for op in operators)

avg_lines_hr = round(total_lines / (total_time / 60), 2)
avg_qty_hr = round(total_qty / (total_time / 60), 2)

print(f"\n📈 Team Average Lines/hr: {avg_lines_hr}")
print(f"📈 Team Average Qty/hr: {avg_qty_hr}")

# Highlight top 3 performers by quantity/hr
sorted_ops = sorted(operators, key=lambda x: x["quantity_actioned"] / (x["work_time"] / 60), reverse=True)
print("\n🏅 Top 3 Performers:")
for op in sorted_ops[:3]:
    print(f"{op['name']} - {round(op['quantity_actioned'] / (op['work_time'] / 60), 2)} Qty/hr")
```
